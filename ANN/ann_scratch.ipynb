```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def ReLU(A):
    return np.maximum(0,A)

def Sigmoid(A):
    return 1/(1+np.exp(-A))

def ForwardPass(x,w1,b1,w2,b2):
    H1 = np.dot(w1, x) + b1
    H1 = ReLU(H1)
    H2 = np.dot(w2, H1) + b2
    H2 = Sigmoid(H2)

    return H2
```
