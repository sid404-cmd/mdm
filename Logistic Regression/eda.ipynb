```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import scipy
import os

data = pd.read_csv("heart.csv")
```

HEATMAP

```python
corr_mat = data.corr()

plt.figure(figsize=(12,8), dpi=300)
sns.heatmap(corr_mat, annot=True, cmap="coolwarm", linewidths=0.5, fmt=".2f")
# plt.savefig("Heatmap")

sns.clustermap(corr_mat, annot=True, cmap="coolwarm", linewidths=0.5, fmt=".2f")
# plt.savefig("Clustermap")

plt.show()
```

```python
features_1 = [col for col in data.columns if col in ["thalach", "slope", "cp"]]
```

Histogram with KDE



```python
plt.figure(figsize=(25,15), dpi = 300)

for i, col in enumerate(features_1):
    plt.subplot(3,2,i+1)
    sns.histplot(
        data=data,
        x = col,
        hue=data["target"], 
        kde=True, 
        palette=["red", "blue"], 
        alpha=0.3
    )

plt.tight_layout(pad = 3)
plt.show()
```

```python
plt.figure(figsize=(25,15), dpi = 300)

plt.subplot(1,2,1)
sns.histplot(
    data=data,
    x = "thalach",
    hue = "sex",
    palette=["red", "blue"],
    # element="step"
)

plt.subplot(1,2,2)
sns.histplot(
    data=data,
    x = "thalach",
    hue = "sex",
    palette=["red", "blue"],
    element="step"
)

plt.tight_layout
plt.show()
```

```python
features_1 = [col for col in data.columns if col in ["thalach", "slope", "cp"]]

plt.figure(figsize=(25, 15))

for i, col in enumerate(features_1):

    plt.subplot(3, 2, 2*i + 1)
    sns.histplot(
        data=data,
        x=col,
        hue="target",
        kde=False,
        palette=["red", "blue"],
        alpha=0.3
    )

    plt.title(f"Distribution of {col}", fontsize=14, fontweight="bold")
    plt.xlabel(col, fontsize=17)
    plt.ylabel("Count", fontsize=17)
    plt.tick_params(axis="both", labelsize=15)
    plt.grid(True, linestyle="--", linewidth=0.5)

    plt.subplot(3, 2, 2*i + 2)
    sns.histplot(
        data=data,
        x=col,
        hue="target",
        kde=True,
        palette=["red", "blue"],
        alpha=0.3
    )

    plt.title(f"Distribution of {col} (KDE)", fontsize=14, fontweight="bold")
    plt.xlabel(col, fontsize=17)
    plt.ylabel("Count", fontsize=17)
    plt.tick_params(axis="both", labelsize=15)
    plt.grid(True, linestyle="--", linewidth=0.5)

plt.tight_layout(pad=3)
plt.show()
```
