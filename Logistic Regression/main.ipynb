```python
features = [col for col in data.columns if col in ["trestbps","chol","thalach","oldpeak"] ]
plt.figure(figsize=(25,15), dpi=300)
for i, col in enumerate(features):
    plt.subplot(3, 2, i+1)
    sns.scatterplot(
        data=data,
        x = "age",
        y = col,
    )  
plt.tight_layout(pad=3)
plt.show()
```

```python
plt.figure(figsize=(25,15), dpi=300)
for i, col in enumerate(features):
    plt.subplot(3, 2, i+1)
    sns.regplot(
        data=data,
        x = "age",
        y = col,
    )  
plt.tight_layout(pad=3)
plt.show()
```

```python
plt.figure(figsize=(15, 15), dpi=300)
sns.set_theme(style="darkgrid")

plt.subplot(1, 1, 1)
sns.scatterplot(
    data=data,
    x="age",
    y="chol"
)

plt.subplot(2, 1, 1)
sns.scatterplot(
    data=data,
    x="age",
    y="chol",
    hue="thalach"
)

plt.tight_layout()
plt.show()
```
